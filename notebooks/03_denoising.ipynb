# OpenDiffraction Benchmark: Diffraction Pattern Denoising

This notebook demonstrates diffraction pattern denoising under Poisson counting and Gaussian readout noise.

```python
import numpy as np
import matplotlib.pyplot as plt
from opendiffraction.objects.specimens import make_core_shell_nanoparticle
from opendiffraction.physics.forward import SimulationConfig, simulate_diffraction
from benchmarks.denoising import denoise_diffraction_pattern, evaluate_denoising

rng = np.random.default_rng(42)
obj = make_core_shell_nanoparticle(64, rng=rng)
cfg = SimulationConfig(photon_budget=1e6)
sim = simulate_diffraction(obj, cfg, rng=rng)

noisy = sim.intensity_measured
expected = sim.intensity_expected
denoised = denoise_diffraction_pattern(noisy, method='median_gaussian')
```

```python
metrics = evaluate_denoising(noisy, expected)
print("Denoising Evaluation Metrics:")
for k, v in metrics.items():
    print(f"  {k}: {v:.4f}")
```

```python
fig, axes = plt.subplots(1, 3, figsize=(12, 4))
axes[0].imshow(np.log1p(np.maximum(expected, 0)), cmap='inferno')
axes[0].set_title('Ground Truth Expected $\log(1+\lambda)$')
axes[1].imshow(np.log1p(np.maximum(noisy, 0)), cmap='inferno')
axes[1].set_title('Noisy Observation')
axes[2].imshow(np.log1p(np.maximum(denoised, 0)), cmap='inferno')
axes[2].set_title('Denoised Reconstruction')
for ax in axes:
    ax.axis('off')
plt.tight_layout()
plt.show()
```
