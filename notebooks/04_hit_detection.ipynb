# OpenDiffraction Benchmark: Hit Finding & Frame Classification

This notebook demonstrates hit finding (single hit, off-center hit, multi-particle hit, non-hit) at XFEL beamlines.

```python
import numpy as np
import matplotlib.pyplot as plt
from opendiffraction.variations.profiles_48 import sample_profile_48
from opendiffraction.physics.forward import simulate_diffraction
from benchmarks.hit_detection import classify_hit

profiles = ["V37", "V38", "V40", "V41", "V42"]
labels = ["High Flux Hit", "Low Flux Hit", "Off-Center Hit", "Multi-Particle Hit", "Empty Carrier Non-Hit"]
frames = []

rng = np.random.default_rng(42)
for pid in profiles:
    cfg, amp, phase, meta = sample_profile_48(pid, index=0, total=1, rng=rng)
    sim = simulate_diffraction(amp, cfg, object_phase=phase, rng=rng)
    frames.append(sim.intensity_measured)

print("Hit Classification Results:")
for p_name, frame in zip(labels, frames):
    res = classify_hit(frame)
    print(f"  {p_name:25s} -> Prediction: {res['prediction']:12s} | Energy: {res['total_energy']:10.1f} | Peak: {res['peak_intensity']:8.1f}")
```

```python
fig, axes = plt.subplots(1, 5, figsize=(15, 3))
for ax, p_name, frame in zip(axes, labels, frames):
    ax.imshow(np.log1p(np.maximum(frame, 0)), cmap='plasma')
    ax.set_title(p_name, fontsize=8)
    ax.axis('off')
plt.tight_layout()
plt.show()
```
