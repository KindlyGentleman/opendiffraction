# OpenDiffraction Benchmark: PyTorch Model Training

This notebook provides a complete PyTorch pipeline for training a neural network reconstructor on OpenDiffraction HDF5 shards.

```python
import os
import torch
import torch.nn as nn
from torch.utils.data import DataLoader
from examples.pytorch_dataset import OpenDiffractionDataset
from examples.minimal_training import TinyPhaseRetrievalNet, train_minimal_epoch

shard_path = 'output_quick/data/train_000.h5'
if os.path.exists(shard_path):
    dataset = OpenDiffractionDataset([shard_path])
    loader = DataLoader(dataset, batch_size=4, shuffle=True)
    print(f"Loaded {len(dataset)} training samples.")

    model = TinyPhaseRetrievalNet()
    optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)
    criterion = nn.MSELoss()

    for epoch in range(3):
        loss = train_minimal_epoch(model, loader, optimizer, criterion)
        print(f"Epoch {epoch+1}/3 - Loss: {loss:.5f}")
```
